<h2> Fetching the data from the TMDB url</h2>

```python
import requests
import time
import pandas as pd

url = "https://api.themoviedb.org/3/movie/top_rated"

all_movies = []

session = requests.Session()

for page in range(1,472):

    params = {
        "api_key": "8265bd1679663a7ea12ac168da84d2e8",
        "language": "en-US",
        "page": page
    }

    success = False

    for attempt in range(5):

        try:
            response = session.get(
                url,
                params=params,
                timeout=30
            )

            if response.status_code == 200:

                data = response.json()
                all_movies.extend(data["results"])

                print(f"Page {page}/471 fetched")

                success = True
                break

            elif response.status_code == 429:

                print(f"Rate limit on page {page}. Waiting 10 seconds...")
                time.sleep(10)

            else:

                print(
                    f"Error on page {page}: "
                    f"{response.status_code}"
                )
                break

        except requests.exceptions.ConnectionError:

            print(
                f"Connection error on page {page} "
                f"(attempt {attempt + 1}/5)"
            )

            time.sleep(5)

    if not success:
        print(f"Could not fetch page {page}. Stopping.")
        break

    # Give the server some time
    time.sleep(1)
```

```python
df = pd.DataFrame(all_movies)

print("Total movies:", len(df))
print("Shape:", df.shape)

df = df[["genre_ids", "title", "overview"]]

print(df.head())
df.to_csv("movies.csv", index=False)

print("movies saved successfully!")
```

<h3>3.Checking the HTMl Tags</h3>

```python
html_count = df['overview'].str.contains(
    r'<[^>]+>',
    regex =True,
    na=False
).sum()

print(f'rows containing html:',html_count)
```

<h3>4.Checking the URLS</h3>

```python
import re

url_pattern = r'https?://\s+|www.\.\S+'

url_count = df['overview'].str.contains(
    url_pattern,
    regex=True,
    na=False
).sum()

print("Rows containing URLs:", url_count)
```

<h3>5.Removing the Emojis</h3>

```python
import emoji

df['text_emoji'] = df['overview'].apply(
    lambda x: emoji.demojize(x)
)
```

<h3>6. Removing  Punctuations</h3>

```python
import string

df['text_punctuation'] = df['text_emoji'].apply(
    lambda x:x.translate(
        str.maketrans('','',string.punctuation)
    )
)
```

<h3>7.Tonization</h3>

```python
from nltk.tokenize import word_tokenize

df['tokens'] = df['text_punctuation'].apply(
    word_tokenize
)
```

<h3>8.Stemming</h3>

```python
from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer("english")

df['stemmed'] = df['tokens'].apply(
    lambda tokens:[stemmer.stem(token)
                  for token in tokens
                  ]
)
```
