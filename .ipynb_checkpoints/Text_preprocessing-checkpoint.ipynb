```python
import requests
import time
import pandas as pd

url = "https://api.themoviedb.org/3/movie/top_rated"

all_movies = []

session = requests.Session()

for page in range(1,471):

    params = {
        "api_key": "8265bd1679663a7ea12ac168da84d2e8",
        "language": "en-US",
        "page": page
    }

    success = False

    for attempt in range(5):

        try:
            response = session.get(
                url,
                params=params,
                timeout=30
            )

            if response.status_code == 200:

                data = response.json()
                all_movies.extend(data["results"])

                print(f"Page {page}/471 fetched")

                success = True
                break

            elif response.status_code == 429:

                print(f"Rate limit on page {page}. Waiting 10 seconds...")
                time.sleep(10)

            else:

                print(
                    f"Error on page {page}: "
                    f"{response.status_code}"
                )
                break

        except requests.exceptions.ConnectionError:

            print(
                f"Connection error on page {page} "
                f"(attempt {attempt + 1}/5)"
            )

            time.sleep(5)

    if not success:
        print(f"Could not fetch page {page}. Stopping.")
        break

    # Give the server some time
    time.sleep(1)
```

```python
df = pd.DataFrame(all_movies)

print("Total movies:", len(df))
print("Shape:", df.shape)

df = df[["genre_ids", "title", "overview"]]

print(df.head())
```

```python
import requests
import time

url = "https://api.themoviedb.org/3/movie/top_rated"

params = {
    "api_key": "8265bd1679663a7ea12ac168da84d2e8",
    "language": "en-US",
    "page": 471
}

response = requests.get(url, params=params, timeout=30)

print("Status:", response.status_code)

if response.status_code == 200:
    data = response.json()

    print("Movies on page 471:", len(data["results"]))

    all_movies.extend(data["results"])

    print("Total movies now:", len(all_movies))
else:
    print(response.text)
```
